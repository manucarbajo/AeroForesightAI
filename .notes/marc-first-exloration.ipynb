# First Thoughts
Das Modell soll das soziotechnische System Flugverkehr in eine agentenbasierte Simulation übertragen.

## 1. Akteure

- **ConsumerAgent**: Repräsentiert den reise-/mobilitätsbedürftigen Konsumenten mit Preissensibilität und Umweltbewusstsein
- **ProviderAgent**: Repräsentiert die gewinnorientierten Fluggesellschaften, die Kapazitäten, Preise und SAF-Adoption steuern
- **RegulatoryAgent**: Repräsentiert Politik und Gesellschaft, die durch CO2-Steuern, SAF-Quoten und Flugverbote Rahmenbedingungen setzen
- **TechnologyAgent**: Repräsentiert den technologischen Fortschritt bei alternativen Kraftstoffen und Antrieben

## 2. Zustände und Einflussfaktoren
Die Agenten basieren auf definierten Einflussfaktoren mit gegensätzlichen Zuständen:

- Faktor A (Mobilitätsbedürfnis – ConsumerAgent):
    - A1 (Global Travel Boom): Hohes Wachstum, Fliegen als primäres Fortbewegungsmittel
    - A2 (Stagnation & Restraint): Verzicht durch Wertewandel (z.B. "Flight Shaming")
- Faktor B (Nachhaltigkeit – RegulatoryAgent):
    - B1 (Market-Oriented): Fokus auf Wirtschaftlichkeit, minimaler Klimaschutz
    - B2 (Radical Decarbonization): Strenge Regulierung durch CO2-Steuern und SAF-Quoten
- Faktor C (Technologie – TechnologyAgent):
    - C1 (Durchbruch): SAF günstig verfügbar, E-Flieger marktreif
    - C2 (Stagnation): Keine skalierbaren Alternativen zu Kerosin
- Faktor D (Energiepreise):
    - D1 (Niedrig): Billiges Kerosin begünstigt Status quo
    - D2 (Hoch): Hohe Preise erzwingen Alternativen

## 3. Die Workbench (MarketEnvironment)
Die übergeordnete Klasse MarketEnvironment dient als Experimentierumgebung. Sie:
- koordiniert alle Agenten pro Simulationsschritt
- führt die Cross-Impact-Bilanz (CIB) Matrix mit allen Deskriptoren und deren Wechselwirkungen
- protokolliert den Simulationsstand als agentisches Gesprächsprotokoll
- prüft Szenario-Konsistenz über die CIB-Matrix

## 4. Interaktionen (wechselwirkungen)
Das Modell erzeugt emergentes Verhalten durch verkettete Agentenreaktionen:
- Regulierung → Provider: CO2-Steuern und SAF-Quoten erzwingen Strategieanpassung (Preiserhöhung, Flottenumbau)
- Regulierung → Consumer: Kurzstreckenverbote oder Subventionen lenken Nachfrage um
- Consumer ↔ Provider: Angebot und Nachfrage bedingen sich gegenseitig – sinkende Nachfrage erzwingt Kapazitätsabbau
- Technology → Provider: Technologische Durchbrüche ermöglichen neue Kraftstoffe und effizientere Flotten

Diese Eigendynamik führt die Simulation zu verschiedenen Ausgängen – z.B. "Green Growth" bei Tech-Durchbruch + moderater Regulierung, oder "Forced Transition" bei hohen Energiepreisen + radikaler Dekarbonisierung.

# Szenario Analyse

## Deskriptoren
Deskriptoren sind die **Schlüsselfaktoren**, die den Flugverkehr der Zukunft beeinflussen:

> SAF - Sustainable Aviation Fuel

| ID | Deskriptor | Description |
|-|-|-|
| A | Nachfrageverhalten | Wie entwickelt sich die Reisenachfrage? |
| B | Regulierung | Wie stark greift die Politik ein? |
| C | Technologie | Wie schnell entwickeln sich alternative Antriebe/SAF? |
| D | Energiepreise | Wie entwickeln sich Öl-/Kerosinpreise? |
| E | Gesellschaftlicher Wandel | Wie stark ist das Umweltbewusstsein? |
| F | Wirtschaftslage | Wie stark ist das Umweltbewusstsein? |
| G | Wettbewerb Alternativen | Wie stark konkurrieren Bahn/Hyperloop etc.? |

## States

Für jeden Deskriptor gibt es 3 States:

| Deskriptor | State 1 | State 2 | State 3 |
|-|-|-|-|
| **A: Nachfrage** | A1: Starkes Wachstum (+5%/Jahr) | A2: Stagnation (0%) | A3: Rückgang (-3%/Jahr) |
| **B: Regulierung** | B1: Marktorientiert (minimal) | B2: Moderate CO2-Bepreisung | B3: Radikale Dekarbonisierung |
| **C: Technologie** | C1: Durchbruch (SAF günstig, E-Flieger) | C2: Langsamer Fortschritt | C3: Stagnation |
| **D: Energiepreise** | D1: Niedrig (< 60$/Barrel Öl) | D2: Moderat (60-100$/Barrel Öl) | D3: Hoch (> 150$/Barrel Öl) |
| **E: Gesellschaft** | E1: Flugscham dominant | E2: Gemischt | E3: Fliegen bleibt normal |
| **F: Wirtschaft** | F1: Boom | F2: Moderate Entwicklung | F3: Rezession |
| **G: Alternativen** | G1: Starke Konkurrenz (Bahn etc.) | G2: Moderate Konkurrenz | G3: Schwache Konkurrenz |
